```python
import re
import nltk
import os
import pandas as pd
from symspellpy import SymSpell, Verbosity
```

```python
# Initialize SymSpell for spelling correction
sym_spell = SymSpell(max_dictionary_edit_distance=2, prefix_length=7)
sym_spell.load_dictionary("path_to_frequency_dictionary.txt", term_index=0, count_index=1)

def preprocess(text):
    # Convert to lowercase
    text = text.lower()
    
    # Remove special characters and digits
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    
    # Replace multiple spaces with a single space
    text = re.sub(r'\s+', ' ', text)
    
    # Tokenize using simple regex
    tokens = re.findall(r'\b\w+\b', text)
    
    # Light spelling correction (only for words not in dictionary)
    corrected_tokens = []
    for token in tokens:
        suggestions = sym_spell.lookup(token, Verbosity.CLOSEST, max_edit_distance=2)
        if suggestions:
            corrected_tokens.append(suggestions[0].term)
        else:
            corrected_tokens.append(token)
    
    # Join tokens back into text
    preprocessed_text = ' '.join(corrected_tokens)
    
    return preprocessed_text
```

```python
df = pd.read_csv(os.path.join(os.path.dirname(os.getcwd()), "data", "data_advice_fulltext.csv"))
docs = list(df["text"])

# Apply preprocessing to your texts
preprocessed_texts = [preprocess(text) for text in docs]
```
